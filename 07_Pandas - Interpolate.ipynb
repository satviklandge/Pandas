## Handling Missing Value By - Interpolate() Function
     In pandas, Missing values are a common issue in real-world datasets and can negatively affect data analysis and model performance.
    The interpolate() function in Pandas provides an effective way to fill missing values by estimating them from existing data points.
    Uses various interpolation techniques

```python
import pandas as pd
```

```python
df1 = pd.read_csv('student_result1.csv')
df1
```

### Using interpolation function 

```python
df1.interpolate()
```

### Using interpolate method - "index"

```python
df1.interpolate(method = 'index')
```

### Using interpolate method - "Linear"

```python
df1.interpolate(method = 'linear')
```

### Using interpolate method - "Time"

```python
# Note : one of the col. should be in time - date dtype
df1 = pd.read_csv('student_result1.csv', index_col = 'Student ID')
df1
```

### Using axis Parameter

```python
df1.interpolate(axis = 0)
```
