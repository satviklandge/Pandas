## Fill NAN Value
        Filling NaN (Missing) Values in Pandas
        In Pandas, NaN (Not a Number) represents missing or undefined data.
        Pandas provides flexible methods to fill NaN values using constants (like 0 or "Unknown"), statistical measures (mean, median, mode), or forward/backward propagation
        
        Most Common Way to Fill NAN Value 
         - ffill -> Fill NAN Values with Forward values
         - bfill -> Fill NAN values with Previous Values
         - etc

```python
import pandas as pd
```

```python
df1 = pd.read_csv("Fortune_10-1.csv")
df1
```

### Fill The NAN Value By Dictionary 

```python
df1.fillna({'Industry':'None','Expenses':'None','Growth':'None'})
```

### Fill NAN by method - ffill

```python
df1.fillna(method = 'ffill', limit = 1,inplace = True)
df1
```

### Fill NAN by method - bfill

```python
df1.fillna(method = 'bfill',inplace = False)
df1
```
