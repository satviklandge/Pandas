## Pandas loc and iloc for accessing rows
    In Pandas, loc and iloc are used to select rows and columns from a DataFrame.
    They are two of the most important indexing methods in data analysis.

    - Use loc when you know the row/column name
    - Use iloc when you know the row/column position number

```python
import pandas as pd
```

```python
df1 = pd.read_csv('student_result1.csv')
df1
```

### 1. Using loc Function

###  For accessing rows

```python
df1.loc[2]
```

### Access Record as field set

```python
df1.loc[[4]]
```

###  Accessing multiple rows

```python
df1.loc[0:3]
```

### For accessing multiple selected record 

```python
df1.loc[[0,2,4]]
```

### For accessing multiple selected record with selected multiple col

```python
df1.loc[2:4,['Student ID', 'Study hrs']]
```

###  Getting Records by condition

```python
df1.loc[df1['Percentage']>=70]
```

### 2. Using iloc

```python
df1.iloc[3]
```

### Access record as field Set

```python
df1.iloc[[3]]
```

### Accessing multiple rows

```python
df1.iloc[0:2]
```

### For Accessing multiple seleted record 

```python
df1.iloc[[0, 2,4]]
```
