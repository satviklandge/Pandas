## Several Parameter of pd.read_csv()

```python
import pandas as pd
```

```python
df1 = pd.read_csv("Fortune_10.csv")
df1.head(5)
```

### Read Only a fixed number of rows from the Dataset

```python
df2 = pd.read_csv("Fortune_10.csv", nrows=3)
df2
```

### Read Only a fixed number of columns from the Dataset

```python
df3 = pd.read_csv("Fortune_10.csv", usecols=[1,2,3,4])
df3
```

### Skips specific rows while reading the dataset

```python
df4 = pd.read_csv("Fortune_10.csv", skiprows = 3)
df4.head(3)
```

### Sets one column as the DataFrame index

```python
df5 = pd.read_csv("Fortune_10.csv", index_col= 0)
df5.head(4)
```

### Specifies which row contains column names

```python
df6 = pd.read_csv("Fortune_10.csv", header= 5)
df6.head(5)
```

### Allocate Custom column names

```python
df7 = pd.read_csv("Fortune_10.csv", names=[1,2,3,4,5,6,7,8,9])
df7.head(5)
```
