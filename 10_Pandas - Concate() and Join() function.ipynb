## Pandas concate() and join () Function
    In Pandas, the concat() function is used to combine two or more DataFrames or Series into a single DataFrame.It helps when you want to:
        -Combine multiple datasets
        -Append rows from one DataFrame to another
        -Add columns from different DataFrames
        -Merge data collected from different files (like CSV files
    The join() function in Pandas is used to combine two DataFrames based on their index or a key column.
        -other → The DataFrame to join
        -on → Column to join on (if not joining on index)
        -how → Type of join
        -'left' (default)
        -'right'
        -'inner'
        -'outer'
        -lsuffix / rsuffix → Suffix for overlapping column names

```python
import pandas as pd
```

```python
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([201,202,203,204,205])
s3 = pd.Series([s1,s2])
s3
```

```python
df1 = pd.DataFrame({'ID' : [1,2,3,4,5],
                    'Name': ['Anil','Sunil','Raju','Rakhi','Ravi'],
                    'Class': [5,6,7,8,9]})

df2 = pd.DataFrame({'ID' : [5,6,7,8],
                    'Name' : ['Sunny','Manoj','Jack','Yash'],
                    'Class': [9,10,11,12]})

df3 = pd.concat([df1,df2])
df3
```

### Concate the DataFrame By col.wise with [axis = 1]

```python
df3 = pd.concat([df1,df2],axis = 1)
df3
```

### For Proper index in DF

```python
df3 = pd.concat([df1,df2], ignore_index = True)
df3
```

```python
df3 = pd.concat([df1,df2],axis = 1, join = 'inner')
df3
```

### Split Data in Two Tables 

```python
df3 = pd.concat([df1,df2],axis = 1, keys = ['Table-1','Table-2'])
df3
```
