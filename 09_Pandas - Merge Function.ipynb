## Pandas Merge Function
    In Pandas, merging is used to join two or more DataFrames into one based on common data columns.By using merge operations, we can connect related data efficiently and create meaningful insights.
     It works similar to SQL joins and helps organize and analyze data that is stored in separate tables.

```python
import pandas as pd
```

## -> Basic Merging 

### Creating two sample DataFrames with a common 'ID' column

```python
df1 = pd.DataFrame({'ID': [1,2,3,4,5],'Class_no': [101,102,103,104,105]})
df2 = pd.DataFrame({'ID': [1,2,3,4,5], 'Class_Teacher': ['Nilesh Sir', 'Swapna Mam', 'Rahul Sir', 'Megha Mam', 'Nitu Mam']})
```

```python
# Merging the two DataFrames; it automatically finds the common column 'ID'
df3 = pd.merge(df1,df2)
df3
```

```python
# Explicitly specifying the 'on' parameter to merge on the 'ID' column
df3 = pd.merge(df1,df2, on ='ID')
df3
```

## -> Merging using other parameters

```python
df1 = pd.DataFrame({'ID': [1,2,3,4,5],'Class_no': [101,102,103,104,105]})
df2 = pd.DataFrame({'ID': [1,2,3,6,7], 'Class_Teacher': ['Nilesh Sir', 'Swapna Mam', 'Rahul Sir', 'Megha Mam', 'Nitu Mam']})
```

```python
# Inner Join (default): Keeps only rows with IDs present in both DataFrames
df3 = pd.merge(df1,df2) # By default how = 'inner'
df3
```

```python
# Left Join: Keeps all records from the left table (df_1), adding NaNs where matches are missing
df3 = pd.merge(df1,df2,how = 'left')  # Give complete record of left side table
df3
```

```python
# Right Join: Keeps all records from the right table (df_2)
df3 = pd.merge(df1,df2,how = 'right')  # Give complete record of right side table
df3
```

```python
# Outer Join: Keeps all records from both tables, filling missing data with NaN
df3 = pd.merge(df1,df2,how = 'outer')  # Give complete record of both side table
df3
```

```python
# indicator function ()
df3 = pd.merge(df1,df2,how = 'outer', indicator = False)  
df3
```

```python
# Using 'indicator=True' to show which table (left, right, or both) the record originated from
df3 = pd.merge(df1,df2,how = 'outer', indicator = True)  # Indicate which record is where
df3
```

```python
# Merging based on the index of the DataFrames and adding suffixes to overlapping column names
df3 = pd.merge(df1, df2, left_index = True, right_index = True, suffixes = ['_Table_1', '_Table_2']) 
df3
```
