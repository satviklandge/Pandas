## Pandas NAN value drop
    In Pandas, NaN (Not a Number) represents missing or undefined values in a dataset.
    The Function dropna() method to remove rows or columns containing NaN values, helping you prepare clean and reliable datasets.
    
        Key Parameters :
            - axis=0	: Drops rows (default behavior)
            - axis=1	: Drops columns
            - how='all'	: Drops only if all values are NaN
            - how='any'	: Drops if any NaN is present (default)
            - thresh=2	: Keeps rows/columns with at least 2 non-null values
            - subset	: Drops NaN based on specific columns only
            - inplace=False	: Returns a new DataFrame (default)
            - inplace=True	: Modifies the original DataFrame

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv("Fortune_5.csv")
df1
```

### Extract null value in true/false format

```python
df2 = pd.read_csv("Fortune_10.csv")
```

```python
df2.isnull().any()
```

### Extract Exact number of null value in Dataset

```python
df1.isnull().sum()
```

### Total Number of null value

```python
print ("Total Null Value in Dataset ;",[(df1.isnull().sum().sum())])
```

### Extract not null Values 

```python
df2.notnull().sum()
```

### Total Number of not null values

```python
print ("Total Not Null Value in Dataset ;",[(df2.notnull().sum().sum())])
```

### For droping the nan values

```python
df2.dropna()
```

### Droping NAN Values by Column 

```python
df2.dropna(axis = 1)
```

### If a rows has complete NAN all feature 

```python
df1.dropna(how = 'all')
```

### If a rows has Atleast One NAN all Feature

```python
df1.dropna(how = 'any')
```

```python
df1.dropna(thresh = 4)
```

### Drop the NAN Value from the Selected Column

```python
df1.dropna(subset = ['Profit','Growth'], inplace = True)
df1.isnull().sum()
```

```python
df1.fillna({'Expense' : '123'})
```
